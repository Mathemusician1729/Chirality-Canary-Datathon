```python
from pathlib import Path
import tkinter as tk
from tkinter import filedialog
import pandas as pd

root = tk.Tk()
# root.withdraw()
filename = filedialog.askopenfilename(
    title="Select well jobs Parquet file",
    filetypes=[("Parquet files", "*.parquet *.pq"), ("All files", "*.*")],
)
root.destroy()

if not filename:
    raise FileNotFoundError("No well jobs file was selected.")

dataset_path = Path(filename)
if dataset_path.suffix.lower() not in {".parquet", ".pq"}:
    raise ValueError("The selected file must be a Parquet file.")

well_jobs = pd.read_parquet(dataset_path)
well_jobs.head(10)
```

```python
well_jobs.drop(columns=[
    "formatted_uwi", "well_name", "job_status_1", "job_status_2", "summary", "actual_final_job_cost", "afe_cost_total", "afe_and_supp_cost_total",
], inplace=True)
```
